```python
import pandas as pd
```

```python
s = pd.Series(['X', 'Y', 'T', 'Aaba', 'Baca', 'CABA', None, 'bird', 'horse', 'dog'])
```

```python
upper_case = s.str.upper()
print("Upper case:\n", upper_case, "\n")
```

```python
lower_case = s.str.lower()
print("Lower case:\n", lower_case, "\n")
```

```python
lengths = s.str.len()
print("Length of strings:\n", lengths)
```
